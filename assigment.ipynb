# Counting Words in a File

```python
def counting_words():
    with open("doc.txt","r") as f:
        text = f.read()
    words = text.split()
    return len(words)
print(counting_words())        
```

# Finding and replacing text

```python
def search_replace( old_one , new_one):
    with open("doc.txt","r") as f:
        text = f.read()
    text = text.replace(old_one,new_one)
    with open("doc.txt","w") as f:
        f.write(text)

search_replace("Hellow","Namaste")
```

# Reading a File in Reverse


```python
def reading_reverse():
    with open("doc.txt","r") as f:
        lines = f.readlines()
    for line in reversed(lines):
        print(line.strip())
reading_reverse()            
```

# counting lines ,words,characters 

```python
def file_all():
    with open("doc.txt","r") as f:
        text = f.read()
    lines = text.splitlines()
    words = text.split()
    chars = len(text)
    return f" no.of lines {len(lines)}, \n no.of words {len(words)}, \n no.of characters {chars}"

print(file_all())
    
```

# Fill NaN with Mean

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1,10,(5,3)),columns=['A','B','C'])
df.iloc[1,1]=np.nan
df.iloc[3,2]=np.nan
df.fillna(df.mean(), inplace=True)
print(df)
```

# Data Aggregation

```python

df=pd.DataFrame({
    'Category':np.random.choice(['A','B','C'],10),
    'value':np.random.randint(1,20,10)
})
print(df.groupby('Category')['value'].agg(['sum','mean']))
```

# Group by Category Sales

```python
df=pd.DataFrame({
    'Product':['P1','P2','P3','P4'],
    'Category':['A','B','A','C'],
    'Sales':np.random.randint(10,100,4)
})
print(df.groupby('Category')['Sales'].sum())
```

# Merging Data Frames

```python
import pandas as pd
df1=pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
df2=pd.DataFrame({'ID':[1,2,3],'Score':[90,80,70]})
merged = pd.merge(df1,df2 ,on='ID')
print(merged)
```
